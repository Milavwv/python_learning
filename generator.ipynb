```python
def countdown(num):
  print("starting")
  while num >= 0:
       yield num
       num -=1

cd =countdown(4)
value = next(cd)
print(value)
```

```python
def firstn(n):
  nums =[]
  num=0
  while num < n:
    nums.append(num)
    num +=1
  return nums

def firstn_generator(n):
  num=0
  while num < n :
    yield num
    num +=1

print(firstn(10))
print(sum(firstn_generator(10)))
```

```python
import sys
def firstn(n):
  nums =[]
  num=0
  while num < n:
    nums.append(num)
    num +=1
  return nums

def firstn_generator(n):
  num=0
  while num < n :
    yield num
    num +=1

print(sys.getsizeof(firstn(100000)))
print(sys.getsizeof(firstn_generator(100000)))
```

```python
def fibonacci_gen(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

for num in fibonacci_gen(10):
    print(num)
```
